1. Write a program to read two strings using input() and display them on console. Perform basic string
operations on them.

```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

print("\nFirst string:", str1)
print("Second string:", str2)

print("\nString Operations:")
print("Concatenation:", str1 + str2)
print("Concatenation with space:", str1 + " " + str2)
print("Length of first string:", len(str1))
print("Length of second string:", len(str2))
print("First string in uppercase:", str1.upper())
print("Second string in lowercase:", str2.lower())
print("First string in reverse:", str1[::-1])
print("Second string in reverse:", str2[::-1])

print("Are both strings equal?", str1 == str2)
print("Is first string present in second?", str1 in str2)
```

2. Write a program to ask the user to enter name and date of birth (YYYY-MM-DD) and validate it
using a function. Display the user's current age, print the calendar for the month and year of birth, and
print a message stating the year the user turns 100.

```python
from datetime import datetime
import calendar


def validate_dob(dob):
    try:
        return datetime.strptime(dob, "%Y-%m-%d")
    except ValueError:
        return None


name = input("Enter your name: ")
dob_input = input("Enter your date of birth (YYYY-MM-DD): ")

dob = validate_dob(dob_input)

if dob is None:
    print("Invalid date of birth!")
else:
    today = datetime.today()

    age = today.year - dob.year

    if (today.month, today.day) < (dob.month, dob.day):
        age -= 1

    year_100 = dob.year + 100

    print("\nName:", name)
    print("Current age:", age)
    print("You will turn 100 in the year:", year_100)

    print("\nCalendar of birth month:")
    print(calendar.month(dob.year, dob.month))
```

3. Write a program that allows the user to enter 10 numbers and displays the largest and smallest odd
and even numbers among them, along with the count of odd and even numbers.

```python
numbers = []

print("Enter 10 numbers:")

for i in range(10):
    num = int(input(f"Enter number {i + 1}: "))
    numbers.append(num)

odd_numbers = []
even_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("\nResult:")

print("Entered Numbers: ", numbers)
print("Count of odd numbers:", len(odd_numbers))
print("Count of even numbers:", len(even_numbers))

if odd_numbers:
    print("Smallest odd number:", min(odd_numbers))
    print("Largest odd number:", max(odd_numbers))
else:
    print("No odd numbers entered.")

if even_numbers:
    print("Smallest even number:", min(even_numbers))
    print("Largest even number:", max(even_numbers))
else:
    print("No even numbers entered.")
```

4. Write a program to implement a menu-driven calculator that performs basic arithmetic operations
(addition, subtraction, multiplication, division) and includes advanced operations like square root,
power, and modulus. Implement error handling when user enters invalid input.

```python
import math


while True:
    print("\nMENU DRIVEN CALCULATOR:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Square Root")
    print("6. Power")
    print("7. Modulus")
    print("8. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 8:
            print("Calculator exited.")
            break

        if choice in [1, 2, 3, 4, 6, 7]:
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))

            if choice == 1:
                print(f"Result of {a} + {b}: {a + b}")

            elif choice == 2:
                print(f"Result of {a} - {b}: {a - b}")

            elif choice == 3:
                print(f"Result of {a} * {b}: {a * b}")

            elif choice == 4:
                if b == 0:
                    raise ZeroDivisionError("Cannot divide by zero.")
                print(f"Result of {a} / {b}: {a / b}")

            elif choice == 6:
                print(f"Result {b} to the power of {a}: {a ** b}")

            elif choice == 7:
                if b == 0:
                    raise ZeroDivisionError("Cannot find modulus with zero.")
                print(f"Result of {a} % {b}: {a % b}")

        elif choice == 5:
            a = float(input("Enter a number: "))

            if a < 0:
                raise ValueError("Square root of a negative number is not allowed.")

            print(f"Result of Square root of {a}: {math.sqrt(a)}")

        else:
            print("Invalid choice. Please select 1 to 8.")

    except ValueError as e:
        print("Invalid input:", e)

    except ZeroDivisionError as e:
        print("Error:", e)
```

5. Write a program to check whether a number is an Armstrong number, implement it with and without
a user-defined function.

(1) Without user-defined function

```python
# Without user-defined function
num = int(input("Enter a number: "))

original = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total += digit ** digits
    num //= 10

if total == original:
    print(original, "is an Armstrong number.")
else:
    print(original, "is not an Armstrong number.")
```

(2) With user-defined function

```python
# With user-defined function
def is_armstrong(num):
    original = num
    digits = len(str(num))
    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** digits
        num //= 10

    return total == original


num = int(input("Enter a number: "))

if is_armstrong(num):
    print(num, "is an Armstrong number.")
else:
    print(num, "is not an Armstrong number.")
```

6. Write a Python function to print the first n rows of Pascal's triangle.

```python
def pascal_triangle(n):
    for i in range(n):
        value = 1

        # Print spaces for triangle shape
        print(" " * (n - i), end="")

        for j in range(i + 1):
            print(value, end=" ")

            value = value * (i - j) // (j + 1)

        print()


n = int(input("Enter number of rows: "))

if n <= 0:
    print("Enter a positive number.")
else:
    pascal_triangle(n)
```

7. Write a Python function to check whether a given string is a pangram.

```python
def is_pangram(text):
    alphabet = set("abcdefghijklmnopqrstuvwxyz")
    return alphabet.issubset(set(text.lower()))


text = input("Enter a string: ")

if is_pangram(text):
    print(f"The string -> {text} <- is a pangram.")
else:
    print(f"The string -> {text} <- is not a pangram.")
```

8. Write a program to find the HCF and GCD of two numbers using a function.

```python
def hcf(a, b):
    while b != 0:
        a, b = b, a % b

    return abs(a)


num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

result = hcf(num1, num2)

print(f"HCF of {num1} & {num2}:", result)
print(f"GCD of {num1} & {num2}:", result)
```
